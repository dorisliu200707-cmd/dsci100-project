```r
cols <- basketball_tidy |> select(-bid) |> names()

avg <- basketball_tidy |> select(-bid) |>
    map_dbl(mean, na.rm=TRUE)

min <- basketball_tidy |> select(-bid) |>
    map_dbl(min, na.rm=TRUE)

max <- basketball_tidy |> select(-bid) |>
    map_dbl(max, na.rm=TRUE)

basketball_4.1 <- tibble(variables = cols, mean = avg, minimum=min, maximum=max)
basketball_4.1
```

Visualization 1: My research question is whether `reg_wins_pct` and `total_wins_pct` can predict `bid`. This scatterplot shows the relationship between the two predictors on the x and y axes, with colors representing the response variable, `bid`. The two predictors show a strong positive relationship. This visualization is relevant because it helps evaluate whether the predictors are informative for knn classification and whether they can effectively predict `bid` by examining the overlap between colored points. Although red (at-large) and green (auto) points are separated in some regions, they are mixed in others, suggesting that knn could have difficulty distinguishing bid categories in those overlapping regions.

```r
options(repr.plot.width = 12, repr.plot.height = 6)
visualization1 <- ggplot(basketball_tidy, aes(x=total_wins_pct, y=reg_wins_pct)) +
    geom_point(aes(color=bid)) +
    labs(x="Total Win Percentage", y="Regional Win Percentage", color = "Bid Category",
        title = "Relationship Between Regional Win Percentage and Total Win Percentage") +
    theme(text=element_text(size=20))
visualization1
```

Visualization 2: This histogram shows the distribution of `total_wins_pct` across the two `bid` categories (auto and at-large). This visualization is relevant because it helps evaluate whether `total_wins_pct` is a strong predictor of `bid` by examining how much the distributions overlap. Less overlap suggests that the predictor can better distinguish between bid categories, while greater overlap makes knn classification more challenging. In this plot, although some distributions overlap, at-large observations are more concentrated around 60–75%, while auto observations are more widely distributed. This suggests that `total_wins_pct` is still useful for knn to predict `bid` but has limited predictive ability on its own.

```r
options(repr.plot.width = 12, repr.plot.height = 6)
visualization2 <- ggplot(basketball_tidy, aes(x=total_wins_pct, fill=bid)) +
    geom_histogram(position="identity", alpha = 0.5) +
    labs(x="Total Win Percentage", y="Count", fill = "Bid Category",
        title="Total Win Percentage Distribution Across Bid Categories") +
    theme(text=element_text(size=20))
visualization2
```
